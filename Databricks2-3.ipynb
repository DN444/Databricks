```python
from pyspark.sql import functions as F

delta_path = "/Volumes/workspace/ecommerce/ecommerce_data/oct_delta"
events = spark.read.format("delta").load(delta_path)
safe_price = F.when(
    (F.col("price").isNotNull()) & 
    (F.col("price") != "") & 
    (F.length(F.trim(F.col("price"))) > 0) &
    (F.col("price").rlike("^[0-9]+\\.?[0-9]*$")),
    F.regexp_replace(F.col("price"), "[^0-9.]", "").cast("double")
).otherwise(None)

features_df = events.groupBy("user_id").agg(
    F.count("*").alias("total_events"),
    F.count(F.when(F.col("event_type") == "purchase", 1)).alias("purchases"),
    F.sum(safe_price).alias("total_spent"),
    F.avg(safe_price).alias("avg_price"),
    F.countDistinct("product_id").alias("unique_products"),
    F.countDistinct("user_session").alias("sessions"),
    F.min("event_time").alias("first_event_date_raw"),
    F.max("event_time").alias("last_event_date_raw")
).orderBy(F.desc("total_events"))
print("✅ Features created")
features_df.show(10, truncate=False)
features_df.printSchema()
```

```python
silver_path = "/Volumes/workspace/ecommerce/ecommerce_data/silver/user_features"
final_features = features_df.dropDuplicates(["user_id"])

final_features.write.format("delta") \
  .mode("overwrite") \
  .option("overwriteSchema", "true") \
  .save(silver_path)

print(f"✅ Silver layer saved: {silver_path}")
print(f"📊 Unique users: {spark.read.format('delta').load(silver_path).count()}")
spark.sql(f"OPTIMIZE delta.`{silver_path}` ZORDER BY (user_id)")
print("✅ Optimized!")
```

```python
def create_user_features(bronze_path, event_type_filter="all"):
    """Modular feature engineering function"""
    print(f"🎯 Processing: {bronze_path}")
    
    events = spark.read.format("delta").load(bronze_path)
    
    safe_price = F.when(
        (F.col("price").isNotNull()) & 
        (F.col("price") != "") & 
        (F.length(F.trim(F.col("price"))) > 0) &
        (F.col("price").rlike("^[0-9]+\\.?[0-9]*$")),
        F.regexp_replace(F.col("price"), "[^0-9.]", "").cast("double")
    ).otherwise(None)
    if event_type_filter != "all":
        events = events.filter(F.col("event_type") == event_type_filter)
    
    features_df = events.groupBy("user_id").agg(
        F.count("*").alias("total_events"),
        F.count(F.when(F.col("event_type") == "purchase", 1)).alias("purchases"),
        F.sum(safe_price).alias("total_spent"),
        F.avg(safe_price).alias("avg_price"),
        F.countDistinct("product_id").alias("unique_products"),
        F.countDistinct("user_session").alias("sessions"),
        F.min("event_time").alias("first_event_date"),
        F.max("event_time").alias("last_event_date")
    ).orderBy(F.desc("total_events"))
    
    return features_df

bronze_path = dbutils.widgets.get("bronze_path")
event_type_filter = dbutils.widgets.get("event_type_filter")

features_df = create_user_features(bronze_path, event_type_filter)
print("✅ Features generated!")
features_df.show(5, truncate=False)
```

```python
def save_silver_layer(features_df, env="prod"):
    """Save features with environment-specific paths"""
    env = dbutils.widgets.get("environment")
    silver_path = f"/Volumes/workspace/ecommerce/ecommerce_data/silver/user_features_{env}"
    
    print(f"💾 Saving to: {silver_path}")
    
    final_features = features_df.dropDuplicates(["user_id"])
    
    final_features.write.format("delta") \
      .mode("overwrite") \
      .option("overwriteSchema", "true") \
      .save(silver_path)
    
    spark.sql(f"OPTIMIZE delta.`{silver_path}`")
    
    print(f"✅ Saved {spark.read.format('delta').load(silver_path).count()} unique users")
    return silver_path

environment = dbutils.widgets.get("environment")
silver_path = save_silver_layer(features_df, environment)
```

```python
def validate_silver_quality(silver_path):
    """Comprehensive quality checks"""
    user_features = spark.read.format("delta").load(silver_path)
    
    print("🏆 PRODUCTION QUALITY REPORT")
    print(f"Path: {silver_path}")
    stats = user_features.select(
        F.count("*").alias("total_users"),
        F.sum("total_events").alias("total_events"),
        F.sum("purchases").alias("total_purchases"),
        F.round(F.avg("total_spent"), 2).alias("avg_spent")
    )
    stats.show(truncate=False)
    quality = user_features.select(
        F.count(F.when(F.col("purchases") > 0, 1)).alias("buyers"),
        F.count(F.when(F.col("total_spent").isNull(), 1)).alias("null_spent"),
        F.approx_count_distinct("user_id").alias("unique_users")
    )
    quality.show(truncate=False)
    
    return user_features

user_features = validate_silver_quality(silver_path)
```
