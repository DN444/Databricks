```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.ecommerce_bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.ecommerce_silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.ecommerce_gold")
base_vol = "/Volumes/workspace/ecommerce/ecommerce_data"

raw_csv = f"{base_vol}/2019-Oct.csv"

bronze_path = f"{base_vol}/delta/bronze/events"
silver_path = f"{base_vol}/delta/silver/events"
gold_path   = f"{base_vol}/delta/gold/product_perf"
```

```python
bronze = (spark.read.format("csv")
          .option("header", "true")
          .option("inferSchema", "true")
          .load(raw_csv)
          .withColumn("ingestion_ts", F.current_timestamp())
          .withColumn("source_file", F.lit(raw_csv)))

(bronze.write
 .format("delta")
 .mode("overwrite")
 .save(bronze_path))

print("Bronze written to:", bronze_path)
```

```python
(bronze.write
 .format("delta")
 .mode("overwrite")
 .saveAsTable("workspace.ecommerce_bronze.events"))
```

```python
bronze_df = spark.read.format("delta").load(bronze_path)

silver = (bronze_df
          .withColumn("event_ts", F.to_timestamp("event_time"))
          .withColumn("event_date", F.to_date("event_ts"))
          .withColumn("price", F.col("price").cast("double"))
          .filter(F.col("event_ts").isNotNull())
          .filter(F.col("event_type").isNotNull())
          .filter(F.col("user_session").isNotNull())
          .filter((F.col("price").isNull()) | ((F.col("price") > 0) & (F.col("price") < 10000)))
          .dropDuplicates(["user_session", "event_time", "event_type", "product_id"])
          .withColumn(
              "price_tier",
              F.when(F.col("price").isNull(), F.lit("unknown"))
               .when(F.col("price") < 10, F.lit("budget"))
               .when(F.col("price") < 50, F.lit("mid"))
               .otherwise(F.lit("premium"))
          )
         )

(silver.write
 .format("delta")
 .mode("overwrite")
 .save(silver_path))

print("Silver written to:", silver_path)
```

```python
(silver.write
 .format("delta")
 .mode("overwrite")
 .saveAsTable("workspace.ecommerce_silver.events"))
```

```python
silver_df = spark.read.format("delta").load(silver_path)

product_perf = (
    silver_df.groupBy("product_id")
    .agg(
        F.countDistinct(F.when(F.col("event_type") == "view", F.col("user_id"))).alias("unique_viewers"),
        F.countDistinct(F.when(F.col("event_type") == "purchase", F.col("user_id"))).alias("unique_purchasers"),
        F.round(F.sum(F.when(F.col("event_type") == "purchase", F.col("price")).otherwise(F.lit(0.0))), 2).alias("revenue")
    )
    .withColumn(
        "conversion_rate_pct",
        F.when(F.col("unique_viewers") == 0, F.lit(0.0))
         .otherwise(F.round((F.col("unique_purchasers") / F.col("unique_viewers")) * 100, 4))
    )
)

(product_perf.write
 .format("delta")
 .mode("overwrite")
 .save(gold_path))

print("Gold written to:", gold_path)

display(product_perf.orderBy(F.col("revenue").desc()).limit(20))
```

```python
(product_perf.write
 .format("delta")
 .mode("overwrite")
 .saveAsTable("workspace.ecommerce_gold.product_perf"))
```

```python
session_conv = (
    events2.groupBy("user_session", "is_weekend")
           .agg(
               F.max(F.when(F.col("event_type") == "purchase", F.lit(1)).otherwise(F.lit(0))).alias("converted")
           )
)

display(session_conv.groupBy("is_weekend", "converted").count().orderBy("is_weekend","converted"))
display(session_conv.groupBy("is_weekend").agg(F.avg("converted").alias("conversion_rate")).orderBy("is_weekend"))
```

```python
w_user = Window.partitionBy("user_id").orderBy("event_ts")
w_session = Window.partitionBy("user_session").orderBy("event_ts")

features = (
    events2
    .filter(F.col("event_ts").isNotNull())
    .withColumn("hour", F.hour("event_ts"))
    .withColumn("day_of_week", F.dayofweek("event_date"))  
    .withColumn("price_clean", F.coalesce(F.col("price").cast("double"), F.lit(0.0)))
    .withColumn("price_log", F.log(F.col("price_clean") + F.lit(1.0)))

    .withColumn("first_event_ts_user", F.first("event_ts").over(w_user))
    .withColumn("time_since_first_event_user_sec",
                F.unix_timestamp("event_ts") - F.unix_timestamp("first_event_ts_user"))

    .withColumn("event_index_in_session", F.row_number().over(w_session))
    .withColumn("prior_event_ts_session", F.lag("event_ts", 1).over(w_session))
    .withColumn("time_since_prev_event_session_sec",
                F.unix_timestamp("event_ts") - F.unix_timestamp("prior_event_ts_session"))
)

display(features.select(
    "user_id","user_session","event_ts","event_type",
    "hour","day_of_week","price_clean","price_log",
    "event_index_in_session","time_since_prev_event_session_sec",
    "time_since_first_event_user_sec"
).limit(10))
```

```python
import mlflow
import mlflow.sklearn
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
exp_path = "/Users/ark4nandi@gmail.com/ecommerce_regression"

mlflow.set_experiment(exp_path)
print("Using experiment:", exp_path)
```

```python
pdf = spark.table("workspace.ecommerce_gold.product_perf").toPandas()

pdf = pdf.replace([np.inf, -np.inf], np.nan).dropna(subset=["unique_viewers", "unique_purchasers", "revenue"])
pdf[["unique_viewers", "unique_purchasers", "revenue"]] = pdf[["unique_viewers", "unique_purchasers", "revenue"]].astype(float)

pdf.head()
```

```python
X = pdf[["unique_viewers", "unique_purchasers"]]
y = pdf["revenue"]

test_size = 0.2
random_state = 42

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=test_size, random_state=random_state
)
```

```python
models = {
    "linear": LinearRegression(),
    "decision_tree": DecisionTreeRegressor(max_depth=5, random_state=random_state),
    "random_forest": RandomForestRegressor(n_estimators=100, random_state=random_state, n_jobs=-1),
}

results = []

for name, model in models.items():
    with mlflow.start_run(run_name=f"{name}_reg_v1"):
        mlflow.log_param("model_type", name)
        mlflow.log_param("features", "unique_viewers,unique_purchasers")
        mlflow.log_param("test_size", test_size)
        mlflow.log_param("random_state", random_state)

        if hasattr(model, "max_depth"):
            mlflow.log_param("max_depth", getattr(model, "max_depth"))
        if hasattr(model, "n_estimators"):
            mlflow.log_param("n_estimators", getattr(model, "n_estimators"))

        model.fit(X_train, y_train)
        preds = model.predict(X_test)

        r2 = r2_score(y_test, preds)
        rmse = mean_squared_error(y_test, preds)
        mae = mean_absolute_error(y_test, preds)

        mlflow.log_metric("r2", float(r2))
        mlflow.log_metric("rmse", float(rmse))
        mlflow.log_metric("mae", float(mae))

        mlflow.sklearn.log_model(model, "model")

        run_id = mlflow.active_run().info.run_id
        results.append((name, run_id, r2, rmse, mae))

        print(f"{name}: R2={r2:.4f}, RMSE={rmse:.2f}, MAE={mae:.2f}, run_id={run_id}")
```

```python
best = sorted(results, key=lambda x: x[2], reverse=True)[0]
best_name, best_run_id, best_r2, best_rmse, best_mae = best

print("Best model (by R2):", best_name)
print("Best run_id:", best_run_id)
print(f"R2={best_r2:.4f}, RMSE={best_rmse:.2f}, MAE={best_mae:.2f}")

display(pd.DataFrame(results, columns=["model", "run_id", "r2", "rmse", "mae"]).sort_values("r2", ascending=False))
```

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression as SparkLR
from pyspark.ml.regression import DecisionTreeRegressor as SparkDT
from pyspark.ml.regression import RandomForestRegressor as SparkRF
from pyspark.ml.evaluation import RegressionEvaluator

spark_df = spark.table("workspace.ecommerce_gold.product_perf") \
    .select(
        F.col("unique_viewers").cast("double").alias("unique_viewers"),
        F.col("unique_purchasers").cast("double").alias("unique_purchasers"),
        F.col("revenue").cast("double").alias("label")
    ) \
    .na.drop()

train_df, test_df = spark_df.randomSplit([0.8, 0.2], seed=42)

assembler = VectorAssembler(inputCols=["unique_viewers", "unique_purchasers"], outputCol="features")
evaluator = RegressionEvaluator(labelCol="label", predictionCol="prediction", metricName="r2")
```

```python
import os
os.environ["MLFLOW_DFS_TMP"] = "/Volumes/workspace/ecommerce/ecommerce_data/mlflow_tmp"
```

```python
spark_models = {
    "spark_lr": SparkLR(featuresCol="features", labelCol="label"),
    "spark_dt": SparkDT(featuresCol="features", labelCol="label", maxDepth=5),
    "spark_rf": SparkRF(featuresCol="features", labelCol="label", numTrees=100),
}

spark_results = []

for name, est in spark_models.items():
    with mlflow.start_run(run_name=f"{name}_pipeline_v1"):
        mlflow.log_param("model_type", name)
        mlflow.log_param("features", "unique_viewers,unique_purchasers")
        if name == "spark_dt":
            mlflow.log_param("maxDepth", 5)
        if name == "spark_rf":
            mlflow.log_param("numTrees", 100)

        pipeline = Pipeline(stages=[assembler, est])
        fitted = pipeline.fit(train_df)

        pred = fitted.transform(test_df)
        r2 = evaluator.evaluate(pred)

        mlflow.log_metric("r2", float(r2))

        mlflow.spark.log_model(fitted, "model",dfs_tmpdir="/Volumes/workspace/ecommerce/ecommerce_data/mlflow_tmp")

        run_id = mlflow.active_run().info.run_id
        spark_results.append((name, run_id, r2))
        print(f"{name}: R2={r2:.4f}, run_id={run_id}")
```

```python
best_spark = sorted(spark_results, key=lambda x: x[2], reverse=True)[0]
print("Best Spark pipeline:", best_spark)
display(pd.DataFrame(spark_results, columns=["model", "run_id", "r2"]).sort_values("r2", ascending=False))
```

```python
import torch
from transformers import pipeline

MODEL_ID = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
REVISION = "714eb0f" 

device = 0 if torch.cuda.is_available() else -1

classifier = pipeline(
    task="sentiment-analysis",
    model=MODEL_ID,
    revision=REVISION,
    device=device
)

reviews = [
    "This product is amazing!",
    "Terrible quality, waste of money"
]

results = classifier(reviews)
results
```
